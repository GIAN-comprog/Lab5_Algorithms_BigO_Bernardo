```python
# [CELL 3] LINEAR GROWTH
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:  
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
# [CELL 4] BIG O CLASSIFICATION
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in[10, 100, 1000, 10000]:
    print(
        f"n={n:4d} | O(1)=1 | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | O(n^2)={n*n:7d}"
    )
```

```python
# [CELL 7] LINEAR SEARCH
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 8] BINARY SEARCH
def binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 10] BUBBLE SORT
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

data = [
    (ID_SUM * 3) % 100, 
    (SEED_DIGIT * 11) % 100, 
    (NAME_LENGTH * 7) % 100, 
    42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python

# [CELL 11] SELECTION SORT
def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

sorted_data, comparisons, swaps = selection_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)  
```

```python
# [CELL 12] INSERTION SORT
def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
# [CELL 13] EXECUTION-TIME MEASUREMENT
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

data_size = 100 + ID_SUM * 10

performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort: {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
# [CELL 15] EXERCISE 1: DATA GENERATION AND INITIALIZATION
LAST_NAME = "BERNARDO"
STUDENT_ID = "TUPM-26-4349"      
FAVORITE_ARTIST = "RADIOHEAD"  

# Metric parameter rules extracted dynamically from manual guidelines
SEED_DIGIT = int(STUDENT_ID[-1])   
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit()) 
ARTIST_LENGTH = len(FAVORITE_ARTIST) 
NAME_LENGTH = len(LAST_NAME)         
ARTIST_SUM = sum(ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha()) 

# Initialize processing datasets
dataset_size = 50 + ID_SUM
base_dataset = [SEED_DIGIT * 5 + i * 3 for i in range(dataset_size)]

# Offset index location using systemic remainder values
rotation_point = (SEED_DIGIT % (dataset_size - 10)) + 5
rotated_dataset = base_dataset[rotation_point:] + base_dataset[:rotation_point]
target_incident = rotated_dataset[(ID_SUM + ARTIST_SUM) % dataset_size] 
```

```python
# [CELL 16] EXERCISE 1: ALGORITHM IMPLEMENTATION
def rotated_binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    comparison_count = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparison_count += 1
        if arr[mid] == target:
            return mid, comparison_count
            
        comparison_count += 1
        if arr[low] <= arr[mid]:
            comparison_count += 1
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            comparison_count += 1
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparison_count

def linear_search_count(arr, target):
    comparison_count = 0
    for item in arr:
        comparison_count += 1
        if item == target:
            break
    return comparison_count
```

```python
# [CELL 17] EXERCISE 1: ASSESSMENT DATA SUMMARY PRINT
found_index, rbs_comparisons = rotated_binary_search(rotated_dataset, target_incident)
ls_comparisons = linear_search_count(rotated_dataset, target_incident)

print("--- Assessment Data: 1. Emergency Dispatch Record Search ---")
print(f"Original Dataset: {base_dataset}")
print(f"Rotated Dataset: {rotated_dataset}")
print(f"Target incident: {target_incident}")
print(f"Linear Search comparison count: {ls_comparisons}")
print(f"Rotated Binary Search comparison count: {rbs_comparisons}")
print(f"Final Output: Target found at index {found_index}")
print("\nTime and space complexity:")
print("- Rotated Binary Search: Time Complexity = O(log n), Space Complexity = O(1)")
print("- Linear Search: Time Complexity = O(n), Space Complexity = O(1)")
print("Observations: Rotated Binary Search drops search checks massively compared to Linear Search.")
```

```python
# [CELL 18] EXERCISE 2: DATA GENERATION
size_A = 30 + SEED_DIGIT
size_B = 25 + ID_SUM % 10

dataset_A = [SEED_DIGIT * 2 + i * 4 for i in range(size_A)]
dataset_B = [ARTIST_LENGTH + i * 3 + NAME_LENGTH for i in range(size_B)]
```

```python
# [CELL 19] EXERCISE 2: MERGE ALGORITHM IMPLEMENTATION
def consolidate_enrollment(list_A, list_B):
    consolidated = []
    i, j = 0, 0
    comparison_count = 0
    
    while i < len(list_A) and j < len(list_B):
        comparison_count += 1
        if list_A[i] <= list_B[j]:
            consolidated.append(list_A[i])
            i += 1
        else:
            consolidated.append(list_B[j])
            j += 1
            
    while i < len(list_A):
        consolidated.append(list_A[i])
        i += 1
    while j < len(list_B):
        consolidated.append(list_B[j])
        j += 1
        
    return consolidated, comparison_count
```

```python
# [CELL 20] EXERCISE 2: ASSESSMENT DATA SUMMARY PRINT
merged_list, total_comps = consolidate_enrollment(dataset_A, dataset_B)

print("--- Assessment Data: 2. University Enrollment Records ---")
print(f"Department A dataset: {dataset_A}")
print(f"Department B dataset: {dataset_B}")
print(f"Consolidated dataset: {merged_list}")
print(f"Number of records before consolidation: {len(dataset_A) + len(dataset_B)}")
print(f"Number of records after consolidation: {len(merged_list)}")
print(f"Comparison count: {total_comps}")
print(f"Ordering verification: {'SUCCESS - List is sorted' if merged_list == sorted(merged_list) else 'FAILED'}")
print("\nTime and space complexity:")
print("- Time Complexity: O(N + M) where elements are parsed concurrently in a single pass.")
print("- Space Complexity: O(N + M) auxiliary space used to house the final collection array.")
print("Final Output: Consolidated collection created without resorting to expensive native .sort() actions.")
```

```python
# [CELL 21] EXERCISE 3: SORTING ALGORITHMS IMPLEMENTATION
def bubble_sort(arr):
    comps, moves = 0, 0
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            comps += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                moves += 3
    return arr, comps, moves

def selection_sort(arr):
    comps, moves = 0, 0
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comps += 1
            if arr[j] < arr[min_idx]:
                min_idx = j
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            moves += 3
    return arr, comps, moves

def insertion_sort(arr):
    comps, moves = 0, 0
    for i in range(1, len(arr)):
        key = arr[i]
        moves += 1
        j = i - 1
        while j >= 0:
            comps += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                moves += 1
                j -= 1
            else:
                break
        arr[j + 1] = key
        moves += 1
    return arr, comps, moves
```

```python
# [CELL 22] EXERCISE 3: PERFORMANCE BENCHMARKING ENGINE
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        data_copy = data.copy()
        start = time.perf_counter()
        result, comps, moves = sort_function(data_copy)
        elapsed = time.perf_counter() - start
        times.append(elapsed)
    avg_time = sum(times) / trials
    return (comps, moves), avg_time

# Configured data allocation limits matching user rule parameters
data_size = 100 + ID_SUM * 10
random_data = [(ID_SUM * 3 + i * 7) % 1000 for i in range(data_size)]
sorted_data = sorted(random_data.copy())
reverse_data = sorted_data.copy()[::-1]
```

```python
# [CELL 23] EXERCISE 3: TESTING EVALUATION AND REPORT WINDOW
conditions = [
    ("Sorted", sorted_data), 
    ("Reverse", reverse_data), 
    ("Random", random_data)
]

print("--- Assessment Data: 3. E-Commerce Order Processing ---")
print(f"Generated order dataset size: {data_size} records.")

for condition_name, current_dataset in conditions:
    b_metrics, b_time = measure(bubble_sort, current_dataset)
    s_metrics, s_time = measure(selection_sort, current_dataset)
    i_metrics, i_time = measure(insertion_sort, current_dataset)
    
    results = [
        ("Bubble Sort", b_metrics, b_time),
        ("Selection Sort", s_metrics, s_time),
        ("Insertion Sort", i_metrics, i_time)
    ]
    
    print(f"\n[Input Condition: {condition_name}]")
    print(f"{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>15}")
    
    for name, metrics, elapsed in results:
        # Correctly extract individual elements out of the tuple
        comparisons = metrics[0]
        movements = metrics[1]
        
        # Now clean formatted values pass correctly without throwing a tuple error
        print(f"{name:<18}{comparisons:>12}{movements:>12}{elapsed:>15.8f}")

print("\nSorting verification: Verified using integrated element tracking copies.")
print("\nBig O Analysis & Case Observations Summary:")
print("- Bubble Sort: Best = O(n) if optimized with checks, Average/Worst = O(n^2). Performs poorly on reverse sets due to mass swaps.")
print("- Selection Sort: Best/Average/Worst = O(n^2). Stable comparisons regardless of data input order, but minimizes elements movement.")
print("- Insertion Sort: Best = O(n) when input is fully pre-sorted, Worst = O(n^2) on reverse datasets.")
print("\nFinal Output: Benchmarking array analysis generation matches performance profile metrics.")

```
